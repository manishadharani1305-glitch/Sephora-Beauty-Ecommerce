## Product Categories

The Sephora Beauty E-Commerce website provides products in different categories to improve customer shopping experience.

- Skincare
- Makeup
- Hair Care
- Fragrance
- Bath & Body

```python
categories = [
    "Skincare",
    "Makeup",
    "Hair Care",
    "Fragrance",
    "Bath & Body"
]

print("Available Categories\n")

for i, category in enumerate(categories, start=1):
    print(f"{i}. {category}")
```

## Sample Products

The following table shows sample beauty products available in the website.

```python
import pandas as pd

products = {
    "Product": ["Lipstick", "Face Wash", "Perfume", "Shampoo", "Moisturizer"],
    "Category": ["Makeup", "Skincare", "Fragrance", "Hair Care", "Skincare"],
    "Price": [899, 599, 2499, 699, 799]
}

df = pd.DataFrame(products)
df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.bar(df["Product"], df["Price"])

plt.title("Product Prices")
plt.xlabel("Products")
plt.ylabel("Price (₹)")

plt.show()
```

## Product Price Analysis

The above chart represents the prices of different beauty products available in the Sephora Beauty E-Commerce website. It helps compare product prices visually.

```python
print("Total Products :", len(df))
print("Average Price  :", df["Price"].mean())
print("Highest Price  :", df["Price"].max())
print("Lowest Price   :", df["Price"].min())
```
