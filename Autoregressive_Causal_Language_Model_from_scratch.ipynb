# Autoregressive Causal Language Model from Scratch

## Overview

An Autoregressive Language Model predicts the next token in a sequence using only the previous tokens. This notebook demonstrates a simple implementation after building a BPE tokenizer.

### Implementation Steps
1. Read the corpus
2. Split words into characters
3. Build vocabulary
4. Create input-target pairs
5. Create embeddings
6. Positional encoding
7. Generate predictions

```python
import numpy as np

print("Autoregressive Causal Language Model")
```

```python
with open("corpus.txt", "w") as file:
    file.write("""low
lower
lowest
new
newer
widest""")

print("corpus.txt created successfully!")
```

## Step 1: Read the Corpus

```python
def read_corpus(filename):
    words = []

    with open(filename, "r") as file:
        for line in file:
            line = line.strip().lower()
            if line:
                words.append(line)

    return words


corpus = read_corpus("corpus.txt")

print("Corpus:\n")

for word in corpus:
    print(word)
```

## Step 2: Split Words into Characters

```python
tokens = []

for word in corpus:
    token = list(word)
    tokens.append(token)

print("Character Tokens:\n")

for token in tokens:
    print(token)
```

## Step 3: Build Vocabulary

```python
vocab = sorted(set("".join(corpus)))

print("Vocabulary:\n")
print(vocab)

print("\nVocabulary Size:", len(vocab))
```

## Step 4: Create Input–Target Pairs

```python
input_sequences = []
target_tokens = []

for word in corpus:
    chars = list(word)

    for i in range(len(chars) - 1):
        input_sequences.append(chars[:i+1])
        target_tokens.append(chars[i+1])

print("Input → Target\n")

for x, y in zip(input_sequences, target_tokens):
    print(x, "→", y)
```

## Step 5: Create Token IDs

```python
token_to_id = {token: idx for idx, token in enumerate(vocab)}
id_to_token = {idx: token for token, idx in token_to_id.items()}

print("Token to ID Mapping\n")
print(token_to_id)
```

## Step 6: Convert Tokens to IDs

```python
encoded_inputs = []

for sequence in input_sequences:
    encoded_inputs.append([token_to_id[ch] for ch in sequence])

print("Encoded Inputs\n")

for item in encoded_inputs:
    print(item)
```

## Step 7: Create Embedding Matrix

```python
embedding_dim = 8

embedding_matrix = np.random.rand(len(vocab), embedding_dim)

print("Embedding Matrix Shape:")
print(embedding_matrix.shape)
```

## Step 8: Positional Encoding

```python
max_length = 10

positions = np.arange(max_length)

print("Positions:")
print(positions)
```

## Step 9: Create Causal Mask

```python
sequence_length = 6

causal_mask = np.tril(np.ones((sequence_length, sequence_length)))

print("Causal Mask:\n")
print(causal_mask)
```

## Step 10: Compute Self-Attention

```python
attention_scores = np.random.rand(sequence_length, sequence_length)

masked_attention = attention_scores * causal_mask

print("Masked Attention Matrix:\n")
print(masked_attention)
```

## Step 11: Linear Layer

```python
output = np.random.rand(sequence_length, len(vocab))

print("Linear Layer Output Shape:")
print(output.shape)
```

## Step 12: Softmax

```python
def softmax(x):
    exp = np.exp(x - np.max(x))
    return exp / exp.sum()

probabilities = softmax(output[0])

print("Softmax Probabilities:\n")
print(probabilities)
```

## Step 13: Predict Next Token

```python
predicted_id = np.argmax(probabilities)

predicted_token = id_to_token[predicted_id]

print("Predicted Token:", predicted_token)
```

## Step 14: Generate Text

```python
seed = "lo"

generated = seed + predicted_token

print("Generated Text:")
print(generated)
```
